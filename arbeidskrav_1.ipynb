### Oppgavetekst
Anta at du skal kjøpe bil. Det står mellom elbil og bensinbil, og du ønsker å sammenlikne de årlige kostnadene ved elbil sammenliknet med bensinbil. Lag et Python-program som beregner og presenterer (viser) de årlige totalkostnadene for elbil og for bensinbil samt årlig kostnadsdifferanse basert på informasjonen gitt nedenfor. Du kan her for enkelhets skyld se bort fra kostnader som renter på billån og verditap (du har da egentlig antatt at slike kostnader er like for elbil og bensinbil).

### Variabler

a er antall kjørte km/år (typisk bilbruk)

```python
a = 10000
```

Forsikring/år i kr. b = elbil. c = bensinbil

```python
b = 5000
c = 7500
```

d er Trafikkforsikringsavgift (likt for både elbil og bensinbil) per dag

```python
d = 8.38
```

Drivstoffbruk: e = Elbil: 0,2 kWh/km. f = Strømpris (antar kun hjemmelading): 2.00 kr/kWh. g = Bensinbil: 1,0 kr/km.

```python
e = 0.2
f = 2
g = 1
```

Bomavgift: h = Elbil: 0,1 kr/km. i = Bensinbil: 0,3 kr/km.

```python
h = 0.1
i = 0.3
```

### Utregning

**Elbil**

```python
elbil = b + (d * 365) + (e*a*f) + (h * a)
```

```python
print ("Elbil er estimert til en total årlig kostnad på",elbil, "kr/år")
```

**Bensinbil**

```python
bensinbil = c + (d * 365) + (g * a) + (i * a)
```

```python
print ("Bensinbil er estimert til en total årlig kostnad på",bensinbil, "kr/år")
```

**Differanse**

```python
print("Bensinbil er dermed estimert til", (bensinbil-elbil),"kr/år dyrere enn elbil.")
```
