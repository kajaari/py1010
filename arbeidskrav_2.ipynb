Lag et program med en funksjon som regner om fra grader til radianer. Programmet skal starte med:
import numpy as np v_grad = float(input('Skriv inn gradtallet:' )) Radiantallet til vinkelen regnes så ut ved følgende formel: v_rad = v_grad*np.pi/180 Resultatet v_rad skrives til skjerm med passende tekst og verdi.
Merk: np.pi er en ferdiglaget funksjon som gir verdien 3.1415....

```python
# Dette programmet importerer numpy for å få tilgang til pi. 
# Brukeren kan legge inn grader som inndata, og de konverteres til radianer som avrundes til 4 desimaler.

import numpy as np
v_grad = float(input("Skriv inn gradtaller:"))
v_rad = round(v_grad*np.pi/180, 4)

print("Det tilsvarer", v_rad, "i radianer.")
```

**Oppg 5.**

Lag et program med en funksjon som tar a og b som inn-argumenter og som så regner ut arealet og «ytre» omkrets til en figur satt sammen av en rettvinklet trekant og en halvsirkel, se figuren under. Med «ytre» omkrets menes samlet lengde av de sorte strekene. Funksjonen skal returnere arealet og «ytre» omkrets, som så skrives til skjerm med passende tekst.



```python
import numpy as np
import math

# Bruker oppgir a og b verdier
a = float(input("Vennligst oppgi a-verdien, som er diameter på halvsikrelen:"))
b = float(input("Vennligst oppgi b-verdien, som er høyden på rettvinklet trekant:"))

# Først regner jeg ut areal og omkrets av sirkel med oppgitte formler
radius = a/2
areal_sirkel = np.pi*radius**2
areal_halvsirkel = areal_sirkel/2
print("Arealet til halvsirkelen er", round(areal_halvsirkel, 4))

omkrets_sirkel = 2*np.pi*radius
omkrets_halvsirkel = omkrets_sirkel/2
print("Omkretsen til halvsirkelen er", round(omkrets_halvsirkel, 4))

# Så regner jeg ut trekanten.
# Jeg tar grunnlinjen (a) * høyden (b) og deler på 2 for å få trekantens areal
areal_trekant = a*b/2
# Jeg bruker pytagoras setning for å finne ut lengden på hypotenus først
hypotenus_trekant = math.sqrt(a**2+b**2)
# OBS: Legger ikke til a her siden vi bruke halvsirkelen
omkrets_trekant = b+hypotenus_trekant

total_omkrets = omkrets_halvsirkel + omkrets_trekant
total_areal = areal_trekant + areal_halvsirkel
print("Omkretsen til figuren er", round(total_omkrets, 4))
print("Arealet til figuren er", round(total_areal, 4))
```
